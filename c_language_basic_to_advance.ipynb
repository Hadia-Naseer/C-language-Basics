```
#include <stdio.h>

int main() {
    int largest, smallest, num;

    printf("Enter five integers:\n");
    scanf("%d %d %d %d %d", &num, &num, &num, &num, &num);

    largest = smallest = num;

    if (num > largest)
        largest = num;
    if (num < smallest)
        smallest = num;

    num = 0;
    while (num < 4) {
        scanf("%d", &num);

        if (num > largest)
            largest = num;
        if (num < smallest)
            smallest = num;
    }

    printf("The smallest integer is %d\n", smallest);
    printf("The largest integer is %d\n", largest);

    return 0;
}
```

```
#include <stdio.h>
float calculate_charges(int hours) {
    if (hours <= 3) {
        return 2.00;
    } else {
        return 2.00 + 0.50 * (hours - 3);
    }
}

int main() {
    int hours[3];
    float charges[3];
    float total_receipts = 0.0;
    printf("Enter parking hours for three customers:\n");
    for (int i = 0; i < 3; i++) {
        scanf("%d", &hours[i]);
        if (hours[i] > 24) {
            printf("Error: Parking hours exceed 24 for customer %d. Please re-enter.\n", i + 1);
            i--;
        }
    }
    for (int i = 0; i < 3; i++) {
        charges[i] = calculate_charges(hours[i]);
        total_receipts += charges[i];
    }

    printf("\nCustomer\tParking Hours\tCharges\n");
    for (int i = 0; i < 3; i++) {
        printf("%d\t\t%d\t\t$%.2f\n", i + 1, hours[i], charges[i]);
    }
    printf("\nTotal Receipts: $%.2f\n", total_receipts);

    return 0;
}
```

```
 #include <stdio.h>
int gcd(int a, int b) {
    if (b == 0) {
        return a;
    } else {
        return gcd(b, a % b);
    }
}

int main() {
    int num1, num2;

    printf("Enter two integers: ");
    scanf("%d %d", &num1, &num2);

    printf("GCD of %d and %d is: %d\n", num1, num2, gcd(num1, num2));

    return 0;
}
```

```
 #include<stdio.h>
//#include<stdlib.h>
void swap(int *a,int *b){
    int hold;
    hold=*a;
    *a=*b;
    *b=hold;
}
int main(){
    int c=7,d=4;
    printf("before swapping a =%d\n b=%d\n",c,d);
    swap(&c,&d);
    printf("now a=%d \n b=%d\n",c,d);
    return 0;
}
```

```
 #include <stdio.h>

int strlen(const char *str) {
    int len = 0;
    while (str[len] != '\0') {
        len++;
    }
    return len;
}

int main() {
    const char *str = "Hello, world!";
    int len = strlen(str);
    printf("Length of string: %d\n", len);
    return 0;
}
```



#String Copy Function



```
 #include <stdio.h>

void strcpy(char *dest, const char *src) {
    while (*src != '\0') {
        *dest = *src;
        dest++;
        src++;
    }
    *dest = '\0';
}

int main() {
    char dest[20];
    const char *src = "Hello, world!";
    strcpy(dest, src);
    printf("Copied string: %s\n", dest);
    return 0;
}
```



#String Concatination



```
    #include <stdio.h>

void strcat(char *dest, const char *src) {
    while (*dest != '\0') {
        dest++;
    }
    while (*src != '\0') {
        *dest = *src;
        dest++;
        src++;
    }
    *dest = '\0';
}

int main() {
    char dest[20];
    const char *src = "Hello, world!";
    strcpy(dest, "Concatenated: ");
    strcat(dest, src);
    printf("Concatenated string: %s\n", dest);
    return 0;}
```

```
 #include <stdio.h>

// Define a structure for a student
struct Student {
    int roll_number;
    char name[50];
    float marks[3];
    float total_marks;
    float average_marks;
};

int main() {
    struct Student students[3];
    for (int i = 0; i < 3; i++) {
        printf("Enter information for student %d:\n", i+1);
        printf("Roll number: ");
        scanf("%d", &students[i].roll_number);
        printf("Name: ");
        scanf("%s", students[i].name);
        printf("Marks for 3 subjects: ");
        for (int j = 0; j < 3; j++) {
            scanf("%f", &students[i].marks[j]);
        }
        students[i].total_marks = students[i].marks[0] + students[i].marks[1] + students[i].marks[2];
        students[i].average_marks = students[i].total_marks / 3;
    }
    printf("\nStudent Information:\n");
    for (int i = 0; i < 3; i++) {
        printf("\nRoll number: %d", students[i].roll_number);
        printf("\nName: %s", students[i].name);
        printf("\nMarks for 3 subjects: ");
        for (int j = 0; j < 3; j++) {
            printf("%.2f ", students[i].marks[j]);
        }
        printf("\nTotal marks: %.2f", students[i].total_marks);
        printf("\nAverage marks: %.2f", students[i].average_marks);
    }

    return 0;
}
```
